# Analisis Multivariat: Prediksi Dropout Mahasiswa
Notebook ini berisi analisis multivariat untuk melihat hubungan antar banyak variabel sekaligus terhadap status mahasiswa (Dropout/Graduate).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Asumsi dataset sudah dimuat ke dalam dataframe bernama 'df'
# Jika membaca dari file baru, gunakan kode di bawah ini:
# df = pd.read_csv('dataset.csv', sep=';') # Sesuaikan nama file

# Menampilkan 5 data teratas
display(df.head())
```

```python
# 1. Correlation Heatmap (Korelasi Multivariat antar fitur numerik)
# Membantu melihat variabel mana saja yang saling berkorelasi tinggi atau memiliki multikolinearitas.
plt.figure(figsize=(16, 12))

numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()

sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Heatmap Korelasi Variabel Numerik', fontsize=16)
plt.show()
```

```python
# 2. Pairplot Multivariat (Distribusi beberapa variabel sekaligus terhadap Target)
selected_cols = ['Curricular units 2nd sem (grade)', 'Curricular units 2nd sem (approved)', 'Age at enrollment', 'Unemployment rate', 'Target']

cols_to_plot = [col for col in selected_cols if col in df.columns]

if len(cols_to_plot) > 1:
    sns.pairplot(df[cols_to_plot], hue='Target', palette='Set1', diag_kind='kde', corner=True)
    plt.suptitle('Pairplot Variabel Pilihan berdasarkan Target', y=1.02, fontsize=14)
    plt.show()
```

```python
# 3. Analisis Feature Importance (Pentingnya Fitur) menggunakan Random Forest
# Analisis ini melihat kombinasi (multivariat) seluruh fitur untuk menentukan fitur mana yang paling kuat memprediksi Target.
from sklearn.ensemble import RandomForestClassifier

df_clean = df.dropna()

if 'Target' in df_clean.columns:
    X = df_clean.drop('Target', axis=1)
    y = df_clean['Target']

    X = pd.get_dummies(X, drop_first=True)

    rf = RandomForestClassifier(n_estimators=100, random_state=42)
    rf.fit(X, y)

    importances = rf.feature_importances_
    indices = np.argsort(importances)[::-1][:15]

    plt.figure(figsize=(12, 6))
    plt.title('Top 15 Feature Importances untuk Prediksi Dropout (Random Forest)', fontsize=14)
    plt.bar(range(15), importances[indices], align='center', color='skyblue')
    plt.xticks(range(15), [X.columns[i] for i in indices], rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
else:
    print("Kolom 'Target' tidak ditemukan di dataset.")
```
