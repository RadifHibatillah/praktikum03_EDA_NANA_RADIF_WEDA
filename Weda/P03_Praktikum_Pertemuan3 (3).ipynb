## Bagian D — Menandai data pribadi

| Kategori | Artinya |
|---|---|
| **pengenal langsung** | Sendirian sudah menunjuk ke satu orang |
| **quasi-identifier** | Sendirian aman, gabungannya menunjuk ke satu orang |
| **data spesifik** | Bocornya merugikan lebih berat (kesehatan, keuangan pribadi, data anak) |
| **bukan data pribadi** | Tidak menunjuk ke orang; inilah bahan analisis |

```python
# TODO D1 — susun tabel penandaan Anda sendiri
# Daftar dibuat berdasarkan nama kolom yang benar-benar tersedia.
# Kolom yang tidak ada akan diabaikan agar kode tetap dapat dijalankan.

PENANDAAN = {}

def tambah_penandaan(nama, kategori, tindakan):
    if nama in df.columns:
        PENANDAAN[nama] = (kategori, tindakan)

# Pengenal langsung: dapat menunjuk/mengidentifikasi satu entitas secara langsung.
for nama in [
    'id_peserta', 'nama_peserta', 'nik', 'no_hp', 'nomor_telepon',
    'email', 'alamat'
]:
    tambah_penandaan(nama, 'pengenal langsung', 'hapus')

# Nama usaha dapat dicari secara publik sehingga diperlakukan sebagai quasi-identifier.
tambah_penandaan('nama_usaha', 'quasi-identifier', 'generalisasi')

# Data yang masih diperlukan untuk analisis tetapi perlu dipertimbangkan risikonya.
for nama in [
    'tanggal_lahir', 'kelurahan', 'kecamatan', 'jenis_kelamin',
    'pendidikan_terakhir'
]:
    tambah_penandaan(nama, 'quasi-identifier', 'generalisasi')

# Data spesifik usaha/ekonomi tidak langsung mengidentifikasi seseorang.
for nama in [
    'omzet_usaha_bulanan', 'jumlah_karyawan', 'lama_usaha_tahun',
    'modal_usaha', 'kategori_usaha', 'kehadiran_persen',
    'skor_prates', 'skor_pascates', 'tingkat_kepuasan',
    'kanal_pendaftaran', 'jumlah_anak'
]:
    tambah_penandaan(nama, 'data spesifik', 'pertahankan')

tanda = pd.DataFrame(
    [
        {'nama_kolom': col, 'kategori': 'bukan data pribadi', 'tindakan': 'pertahankan'}
        for col in df.columns
    ]
)

for nama, (kategori, tindakan) in PENANDAAN.items():
    tanda.loc[tanda['nama_kolom'] == nama, ['kategori', 'tindakan']] = kategori, tindakan

print("Penandaan:")
display(tanda)

print("\nJumlah kolom per kategori:")
print(tanda['kategori'].value_counts())
```

## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS

kolom_hapus = tanda.loc[tanda['tindakan'] == 'hapus', 'nama_kolom'].tolist()

aman = df.drop(columns=kolom_hapus, errors='ignore').copy()

print("Kolom yang dibuang:", len(kolom_hapus))
print("Daftar kolom yang dibuang:", kolom_hapus)
print("Jumlah kolom tersisa:", aman.shape[1])
```

```python
# TODO E2 — HASH

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan

    teks = str(nilai)
    hasil = hashlib.sha256((str(garam) + teks).encode('utf-8')).hexdigest()
    return hasil[:12]

kolom_hash = tanda.loc[tanda['tindakan'] == 'hash', 'nama_kolom'].tolist()

for kolom in kolom_hash:
    aman[kolom] = aman[kolom].apply(lambda x: samarkan(x, GARAM))

print("Kolom yang di-hash:", kolom_hash)
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca

nilai_uji = 'contoh-peserta'

hash_1 = samarkan(nilai_uji, GARAM)
hash_2 = samarkan(nilai_uji, GARAM)

print("Hash pertama :", hash_1)
print("Hash kedua   :", hash_2)
print("Apakah identik?", hash_1 == hash_2)

hash_a = samarkan('peserta-A', GARAM)
hash_b = samarkan('peserta-B', GARAM)

print("\nHash peserta-A:", hash_a)
print("Hash peserta-B:", hash_b)
print("Apakah berbeda?", hash_a != hash_b)
```

**Jawaban Anda:**

Jika `GARAM` ikut ter-*commit* ke GitHub, hash tetap tidak langsung menampilkan nilai asli karena yang tersimpan adalah hasil hash, bukan teks aslinya. Namun, perlindungan menjadi lebih lemah karena pihak lain yang mengetahui `GARAM` dapat mencoba menebak nilai asli dari kandidat yang mungkin, sehingga `GARAM` seharusnya tidak disimpan di repositori publik.


```python
# TODO E4 — GENERALISASI

ACUAN = pd.Timestamp('2025-09-01')

# Pastikan tanggal lahir bertipe datetime.
aman['tanggal_lahir'] = pd.to_datetime(aman['tanggal_lahir'], errors='coerce')

# a. Hitung usia pada tanggal acuan.
aman['usia'] = (
    ACUAN.year - aman['tanggal_lahir'].dt.year
    - (
        (ACUAN.month < aman['tanggal_lahir'].dt.month)
        | (
            (ACUAN.month == aman['tanggal_lahir'].dt.month)
            & (ACUAN.day < aman['tanggal_lahir'].dt.day)
        )
    ).astype(int)
)

# b. Bentuk kelompok umur.
bins = [-np.inf, 24, 34, 44, 54, np.inf]
labels = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(
    aman['usia'],
    bins=bins,
    labels=labels,
    right=True
)

# c. Buang tanggal lahir dan kelurahan.
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'], errors='ignore')

# d. Rapikan penulisan kecamatan.
if 'kecamatan' in aman.columns:
    aman['kecamatan'] = (
        aman['kecamatan']
        .astype('string')
        .str.strip()
        .str.replace(r'\s+', ' ', regex=True)
        .str.title()
    )

print("Distribusi kelompok umur:")
print(aman['kelompok_umur'].value_counts(sort=False))

print("\nJumlah kecamatan unik:", aman['kecamatan'].nunique(dropna=True))
print("Kecamatan:", sorted(aman['kecamatan'].dropna().unique()))
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)

def cek_k(data, kolom):
    # Hitung ukuran setiap kombinasi quasi-identifier.
    ukuran_grup = data.groupby(kolom, dropna=False).size()

    hasil = pd.Series({
        'k': int(ukuran_grup.min()) if len(ukuran_grup) else 0,
        'grup berisi 1 orang': int((ukuran_grup == 1).sum()),
        'total grup': int(len(ukuran_grup))
    })

    return hasil
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan

qi_sebelum = [
    'kelurahan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'tanggal_lahir'
]

risiko_sebelum = cek_k(df, qi_sebelum)

print("Risiko sebelum penanganan:")
print(risiko_sebelum)

# Pemeriksaan tambahan sesuai Expected output.
print("\nJumlah baris dengan grup berisi satu orang:",
      risiko_sebelum['grup berisi 1 orang'])
```

**Jawaban Anda:**

Walaupun keempat kolom tersebut tidak memuat nama atau nomor telepon, kombinasinya dapat membentuk ciri yang cukup unik untuk membedakan seseorang dari peserta lain. Jika kombinasi tersebut dibandingkan dengan sumber data lain, seseorang dapat berpotensi diidentifikasi kembali.


```python
# TODO F3 — ukur risiko SESUDAH generalisasi

qi_sesudah = [
    'kecamatan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'kelompok_umur'
]

risiko_sesudah = cek_k(aman, qi_sesudah)

print("Risiko sesudah generalisasi:")
print(risiko_sesudah)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Coba beberapa kombinasi quasi-identifier dan pilih yang pertama mencapai k >= 5.
# Tujuannya mempertahankan informasi sebanyak mungkin, sehingga kita menghindari
# penghapusan kolom jika kombinasi yang lebih sederhana sudah cukup.

kandidat = [
    ['kecamatan', 'jenis_kelamin', 'kelompok_umur'],
    ['kecamatan', 'kelompok_umur'],
    ['jenis_kelamin', 'kelompok_umur'],
    ['kecamatan'],
]

hasil_f4 = []

for kolom in kandidat:
    if all(k in aman.columns for k in kolom):
        metrik = cek_k(aman, kolom)
        hasil_f4.append({
            'kombinasi': kolom,
            'k': metrik['k'],
            'grup_1_orang': metrik['grup berisi 1 orang'],
            'total_grup': metrik['total grup']
        })

hasil_f4 = pd.DataFrame(hasil_f4)
display(hasil_f4)

layak = hasil_f4[hasil_f4['k'] >= 5]

if len(layak) > 0:
    pilihan_f4 = layak.iloc[0]
    print("Kombinasi yang mencapai k >= 5:", pilihan_f4['kombinasi'])
    print("k yang dicapai:", pilihan_f4['k'])
else:
    print("Belum mencapai k >= 5. Generalisasikan kelompok umur atau gabungkan kecamatan kecil menjadi 'Lainnya'.")

# Contoh generalisasi kecamatan berdasarkan frekuensi:
# kecamatan yang jumlahnya kurang dari 5 digabung menjadi 'Lainnya'.
freq_kec = aman['kecamatan'].value_counts(dropna=False)
aman_f4 = aman.copy()
aman_f4['kecamatan_f4'] = aman_f4['kecamatan'].where(
    aman_f4['kecamatan'].map(freq_kec).ge(5),
    'Lainnya'
)

uji_f4 = cek_k(
    aman_f4,
    ['kecamatan_f4', 'jenis_kelamin', 'kelompok_umur']
)

print("\nSetelah kecamatan kecil digabung menjadi 'Lainnya':")
print(uji_f4)
```

**Jawaban Anda:**

Saya mempertahankan informasi jenis kelamin dan kelompok umur, lalu menggeneralisasi kecamatan yang jumlah pesertanya terlalu sedikit menjadi kategori **"Lainnya"**. Nilai **k akhir harus dibaca dari output `uji_f4` setelah kode dijalankan**; informasi yang dikorbankan adalah detail kecamatan kecil, tetapi informasi kelompok umur dan jenis kelamin tetap dipertahankan.


```python
# TODO F5 — sel kecil pada tabel agregat

# a. Tabel silang jumlah peserta menurut kecamatan x kategori_usaha.
tab_silang = pd.crosstab(
    aman['kecamatan'],
    aman['kategori_usaha']
)

print("Tabel silang kecamatan x kategori_usaha:")
display(tab_silang)

# b. Hitung jumlah sel yang terisi tetapi kurang dari 5 peserta.
sel_kecil = ((tab_silang > 0) & (tab_silang < 5)).sum().sum()

print("Jumlah sel bernilai 1–4:", int(sel_kecil))
print("Kesimpulan: sel kecil sebaiknya tidak ditampilkan apa adanya dalam laporan karena dapat meningkatkan risiko identifikasi.")
```

### G3 — Kamus data

Untuk setiap kolom dataset tim, gunakan format berikut dan isi berdasarkan dataset asli. Jangan menebak arti kolom.

| nama_kolom | tipe_data | deskripsi | skala_pengukuran | kategori_data_pribadi | tindakan_penanganan |
|---|---|---|---|---|---|
| `[nama kolom 1]` | `[tipe]` | `[makna kolom]` | `[nominal/ordinal/interval/rasio]` | `[kategori]` | `[hapus/hash/generalisasi/pertahankan]` |
| `[nama kolom 2]` | `[tipe]` | `[makna kolom]` | `[skala]` | `[kategori]` | `[tindakan]` |

**Catatan:** semua kolom harus dicatat, termasuk kolom yang bukan data pribadi. Kategori dan tindakan harus konsisten dengan hasil pemeriksaan risiko pada Bagian D–F.


```python
# G4 — Mount Google Drive, jalankan penanganan data tim, dan periksa nilai k

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import os

# Tentukan path dataset proyek
FILE_PROYEK = '/content/drive/MyDrive/SEMESTER 3/Expolatory D. Analisys/Pertemuan03/data/dataset.csv'

if os.path.exists(FILE_PROYEK):
    df_proyek = pd.read_csv(FILE_PROYEK)
    print(f"Menggunakan dataset proyek: {FILE_PROYEK} ({df_proyek.shape[0]} baris)")
else:
    print(f"Berkas '{FILE_PROYEK}' belum ditemukan di runtime. Membuat data simulasi proyek...")

    df_proyek = pd.DataFrame({
        'Age at enrollment': [18, 22, 28, 35, 42, 19, 21, 23, 31, 36] * 10,
        'Gender': [1, 0, 1, 1, 0, 0, 1, 0, 1, 1] * 10,
        'Marital status': [1, 1, 2, 1, 3, 1, 1, 2, 1, 4] * 10
    })

# 1. Tentukan Quasi-Identifier
quasi_identifiers = ['Age at enrollment', 'Gender', 'Marital status']

# 2. Generalisasi 1: Umur
def generalize_age(age):
    if age < 20:
        return '<20'
    elif 20 <= age <= 24:
        return '20-24'
    elif 25 <= age <= 29:
        return '25-29'
    elif 30 <= age <= 34:
        return '30-34'
    else:
        return '35+'

df_proyek['kelompok_umur'] = df_proyek['Age at enrollment'].apply(generalize_age)

# 3. Generalisasi 2: Status Pernikahan
def generalize_marital(status):
    if status == 1:
        return 'Single'
    else:
        return 'Lainnya'

df_proyek['marital_grouped'] = df_proyek['Marital status'].apply(generalize_marital)

# 4. Hitung Nilai K setelah Generalisasi
quasi_identifiers_final = ['kelompok_umur', 'Gender', 'marital_grouped']

grouped_final = df_proyek.groupby(quasi_identifiers_final).size()

k_final = grouped_final.min()
total_groups = len(grouped_final)
groups_of_one = (grouped_final == 1).sum()

print("\nHASIL PENANGANAN K-ANONYMITY:")
print(f"Quasi-Identifiers: {quasi_identifiers_final}")
print(f"Nilai K akhir: {k_final}")
print(f"Total kelompok: {total_groups}")
print(f"Jumlah kelompok dengan 1 orang: {groups_of_one}")

if k_final >= 5:
    print("STATUS: BERHASIL mencapai target k >= 5.")
else:
    print("STATUS: GAGAL, butuh generalisasi lebih lanjut.")

# 5. Buang kolom asli yang sudah digeneralisasi
aman_df = df_proyek.drop(columns=['Age at enrollment', 'Marital status'])

# 6. Simpan hasil
# aman_df.to_csv('/content/drive/MyDrive/praktikum EDA/pertemuan3/dataset_aman.csv', index=False)
```
