```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'ganti-dengan-kalimat-rahasia-tim'
```

```python
import hashlib
import numpy as np

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam):
    # Jika nilai kosong, kembalikan NaN
    if pd.isna(nilai):
        return np.nan

    # Gabungkan garam + nilai, lalu buat SHA-256
    teks = str(garam) + str(nilai)
    hasil = hashlib.sha256(teks.encode()).hexdigest()

    # Ambil 12 karakter pertama
    return hasil[:12]


# Cari kolom yang diberi tindakan 'hash'
kolom_hash = tanda.loc[
    tanda['tindakan'] == 'hash', 'nama_kolom'
].tolist()

# Terapkan hashing pada kolom tersebut
for kolom in kolom_hash:
    aman[kolom] = aman[kolom].apply(
        lambda nilai: samarkan(nilai, GARAM)
    )

# Tampilkan contoh hasil
print("Kolom yang di-hash:", kolom_hash)
print(aman[kolom_hash].head())
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

```

```python
# a. Hash dua kali nilai yang sama
nilai_sama = 'PESERTA001'

hash_1 = samarkan(nilai_sama, GARAM)
hash_2 = samarkan(nilai_sama, GARAM)

print("Hash pertama :", hash_1)
print("Hash kedua   :", hash_2)
print("Identik      :", hash_1 == hash_2)

# b. Hash dua nilai yang berbeda
nilai_1 = 'PESERTA001'
nilai_2 = 'PESERTA002'

hash_a = samarkan(nilai_1, GARAM)
hash_b = samarkan(nilai_2, GARAM)

print("\nHash nilai 1 :", hash_a)
print("Hash nilai 2 :", hash_b)
print("Berbeda      :", hash_a != hash_b)
```

```python
# TODO E4 — GENERALISASI

ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta pada tanggal acuan
aman['usia'] = (
    ACUAN.year - aman['tanggal_lahir'].dt.year
    - (
        (aman['tanggal_lahir'].dt.month > ACUAN.month) |
        (
            (aman['tanggal_lahir'].dt.month == ACUAN.month) &
            (aman['tanggal_lahir'].dt.day > ACUAN.day)
        )
    ).astype(int)
)

# b. Buat kelompok umur
aman['kelompok_umur'] = pd.cut(
    aman['usia'],
    bins=[-float('inf'), 24, 34, 44, 54, float('inf')],
    labels=['<25', '25-34', '35-44', '45-54', '55+']
)

# c. Buang tanggal_lahir dan kelurahan
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'])

# d. Rapikan penulisan kecamatan
aman['kecamatan'] = aman['kecamatan'].str.strip().str.title()

# Cek hasil
print("Jumlah peserta per kelompok umur:")
print(aman['kelompok_umur'].value_counts().sort_index())

print("\nKelompok umur terbanyak:")
print(aman['kelompok_umur'].value_counts().idxmax())

print("\nJumlah nilai unik kecamatan:", aman['kecamatan'].nunique())

```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

```

```python
def cek_k(data, kolom):
    # Hitung jumlah orang pada setiap grup
    ukuran_grup = data.groupby(kolom, dropna=False).size()

    # Nilai k = ukuran grup terkecil
    k = ukuran_grup.min()

    # Jumlah grup yang hanya berisi 1 orang
    grup_satu_orang = (ukuran_grup == 1).sum()

    # Total jumlah grup
    total_grup = len(ukuran_grup)

    # Kembalikan hasil dalam bentuk Series
    return pd.Series({
        'k (grup terkecil)': k,
        'grup berisi 1 orang': grup_satu_orang,
        'total grup': total_grup
    })
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
```

```python
kolom_f2 = [
    'kelurahan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'tanggal_lahir'
]

hasil_f2 = cek_k(df, kolom_f2)

print(hasil_f2)
```

```python
# G4 — jalankan penanganan pada data tim dan periksa nilai k
# Salin pola Bagian E dan F, ganti dataset dan nama kolomnya.
# Simpan hasilnya ke data/processed/ dan JANGAN unggah data/raw/

# G4 — Mount Google Drive, jalankan penanganan data tim, dan periksa nilai k

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import os

# Tentukan path dataset proyek
FILE_PROYEK = '/content/drive/MyDrive/EDA Praktikum 3/dataset.csv'

if os.path.exists(FILE_PROYEK):
    df_proyek = pd.read_csv(FILE_PROYEK)
    print(f"Menggunakan dataset proyek: {FILE_PROYEK} ({df_proyek.shape[0]} baris)")
else:
    print(f"Berkas '{FILE_PROYEK}' belum ditemukan di runtime. Membuat data simulasi proyek...")

    df_proyek = pd.DataFrame({
        'Age at enrollment': [18, 22, 28, 35, 42, 19, 21, 23, 31, 36] * 10,
        'Gender': [1, 0, 1, 1, 0, 0, 1, 0, 1, 1] * 10,
        'Marital status': [1, 1, 2, 1, 3, 1, 1, 2, 1, 4] * 10
    })

# 1. Tentukan Quasi-Identifier
quasi_identifiers = ['Age at enrollment', 'Gender', 'Marital status']

# 2. Generalisasi 1: Umur
def generalize_age(age):
    if age < 20:
        return '<20'
    elif 20 <= age <= 24:
        return '20-24'
    elif 25 <= age <= 29:
        return '25-29'
    elif 30 <= age <= 34:
        return '30-34'
    else:
        return '35+'

df_proyek['kelompok_umur'] = df_proyek['Age at enrollment'].apply(generalize_age)

# 3. Generalisasi 2: Status Pernikahan
def generalize_marital(status):
    if status == 1:
        return 'Single'
    else:
        return 'Lainnya'

df_proyek['marital_grouped'] = df_proyek['Marital status'].apply(generalize_marital)

# 4. Hitung Nilai K setelah Generalisasi
quasi_identifiers_final = ['kelompok_umur', 'Gender', 'marital_grouped']

grouped_final = df_proyek.groupby(quasi_identifiers_final).size()

k_final = grouped_final.min()
total_groups = len(grouped_final)
groups_of_one = (grouped_final == 1).sum()

print("\nHASIL PENANGANAN K-ANONYMITY:")
print(f"Quasi-Identifiers: {quasi_identifiers_final}")
print(f"Nilai K akhir: {k_final}")
print(f"Total kelompok: {total_groups}")
print(f"Jumlah kelompok dengan 1 orang: {groups_of_one}")

if k_final >= 5:
    print("STATUS: BERHASIL mencapai target k >= 5.")
else:
    print("STATUS: GAGAL, butuh generalisasi lebih lanjut.")

# 5. Buang kolom asli yang sudah digeneralisasi
aman_df = df_proyek.drop(columns=['Age at enrollment', 'Marital status'])

# 6. Simpan hasil
# aman_df.to_csv('/content/drive/MyDrive/praktikum EDA/pertemuan3/dataset_aman.csv', index=False)
```
