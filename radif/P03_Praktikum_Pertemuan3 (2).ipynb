```python
# TODO E1 — HAPUS
# Buat DataFrame baru bernama 'aman' yang sudah membuang semua kolom bertindakan 'hapus'.
# Cetak berapa kolom yang dibuang dan berapa yang tersisa.

# Identifikasi kolom yang harus dihapus dari tabel 'tanda'
kolom_dihapus = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()

# Buat DataFrame 'aman'
aman = df.drop(columns=kolom_dihapus)

print(f"Jumlah kolom yang dibuang: {len(kolom_dihapus)} ({', '.join(kolom_dihapus)})")
print(f"Jumlah kolom tersisa di DataFrame 'aman': {aman.shape[1]}")
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan
    # Gabungkan garam dan nilai sebagai string, lalu lakukan hashing
    input_string = str(garam) + str(nilai)
    hash_object = hashlib.sha256(input_string.encode('utf-8'))
    return hash_object.hexdigest()[:12]

# Terapkan fungsi penyamaran pada kolom 'id_peserta' di DataFrame 'aman'
aman['id_peserta'] = aman['id_peserta'].apply(lambda x: samarkan(x, GARAM))

# Tampilkan perbandingan beberapa baris sebelum dan sesudah
print("Contoh hasil hash pada kolom 'id_peserta' (5 baris pertama):")
print(aman['id_peserta'].head())
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

nilai_a = 'PLT-2025-1309'
nilai_b = 'PLT-2025-1512'

hash_a1 = samarkan(nilai_a, GARAM)
hash_a2 = samarkan(nilai_a, GARAM)
hash_b = samarkan(nilai_b, GARAM)

print(f"Nilai asli A: {nilai_a}")
print(f"Hash pertama A: {hash_a1}")
print(f"Hash kedua A  : {hash_a2}")
print(f"Apakah hash A1 dan A2 identik? {hash_a1 == hash_a2}")
print("-" * 40)
print(f"Nilai asli B: {nilai_b}")
print(f"Hash B        : {hash_b}")
print(f"Apakah hash A dan B berbeda? {hash_a1 != hash_b}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

* **Yang tidak lagi terlindungi:** Keamanan enkripsi *one-way hash* pada kolom `id_peserta` akan runtuh. Siapa pun yang mendapatkan akses ke dataset ter-hash dan kode sumber Anda dapat melakukan serangan pencocokan nilai (*dictionary attack* atau *brute-force*) untuk merekonstruksi kembali ID peserta asli dengan mudah.
* **Yang tetap terlindungi:** ID peserta asli tetap tidak tampil secara langsung/kasat mata (*plain text*) di dataset, sehingga masih cukup aman dari pembaca awam yang hanya melihat data secara sekilas tanpa niat membongkarnya.

```python
# TODO E4 — GENERALISASI
# a. Hitung usia peserta pada tanggal acuan 1 September 2025
# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
# Expected: kelompok terbanyak adalah 35-44; kecamatan punya tepat 10 nilai unik

ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta (dalam tahun)
usia = (ACUAN - aman['tanggal_lahir']).dt.days // 365.25

# b. Buat kolom 'kelompok_umur'
bins = [-np.inf, 24, 34, 44, 54, np.inf]
labels = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(usia, bins=bins, labels=labels)

# c. Buang kolom tanggal_lahir dan kelurahan (jika masih ada)
kolom_dibuang = [col for col in ['tanggal_lahir', 'kelurahan'] if col in aman.columns]
aman = aman.drop(columns=kolom_dibuang)

# d. Rapikan kolom kecamatan agar konsisten (Title Case)
aman['kecamatan'] = aman['kecamatan'].str.title().str.strip()

# Tampilkan hasil verifikasi
print("Jumlah peserta per kelompok umur:")
print(aman['kelompok_umur'].value_counts().sort_index())
print("-" * 40)
print(f"Jumlah kecamatan unik: {aman['kecamatan'].nunique()}")
print("Daftar kecamatan unik:", aman['kecamatan'].unique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

def cek_k(data, kolom):
    # Kelompokkan data berdasarkan kolom quasi-identifier dan hitung jumlah anggotanya
    ukuran_grup = data.groupby(kolom).size()

    k = ukuran_grup.min()
    grup_satu_orang = (ukuran_grup == 1).sum()
    total_grup = len(ukuran_grup)

    return pd.Series({
        'k': k,
        'grup_berisi_1_orang': grup_satu_orang,
        'total_grup': total_grup
    })
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang

kolom_qi_awal = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
hasil_sebelum = cek_k(df, kolom_qi_awal)

print("Analisis Risiko Sebelum Penanganan (Data Asli):")
print(hasil_sebelum)
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

Kombinasi data spesifik seperti kelurahan, jenis kelamin, jenjang pendidikan, dan tanggal lahir yang sangat presisi dapat bertindak sebagai sidik jari unik (*fingerprint*) yang hanya dimiliki oleh satu orang tertentu di dunia nyata. Jika data anonim ini digabungkan (*data linkage*) dengan basis data publik eksternal yang tidak disensor (seperti DPT Pemilu atau media sosial), penyerang dapat dengan mudah memetakan kembali identitas asli pemilik data tersebut secara akurat.

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1

kolom_qi_baru = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
hasil_sesudah = cek_k(aman, kolom_qi_baru)

print("Analisis Risiko Setelah Penanganan (Data Aman):")
print(hasil_sesudah)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Kita akan mencoba mengeluarkan satu kolom dengan kardinalitas tinggi dari QI: 'kecamatan'
# Mari kita uji k-anonymity tanpa kolom 'kecamatan', hanya dengan: ['jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']

kolom_qi_tantangan = ['jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
hasil_tantangan = cek_k(aman, kolom_qi_tantangan)

print("Analisis Risiko Skenario Tantangan (Tanpa Kecamatan):")
print(hasil_tantangan)
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

```python
# G4 — Mount Google Drive, jalankan penanganan data tim, dan periksa nilai k

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import os

# Tentukan path dataset proyek
FILE_PROYEK = ''

if os.path.exists(FILE_PROYEK):
    df_proyek = pd.read_csv(FILE_PROYEK)
    print(f"Menggunakan dataset proyek: {FILE_PROYEK} ({df_proyek.shape[0]} baris)")
else:
    print(f"Berkas '{FILE_PROYEK}' belum ditemukan di runtime. Membuat data simulasi proyek...")

    df_proyek = pd.DataFrame({
        'Age at enrollment': [18, 22, 28, 35, 42, 19, 21, 23, 31, 36] * 10,
        'Gender': [1, 0, 1, 1, 0, 0, 1, 0, 1, 1] * 10,
        'Marital status': [1, 1, 2, 1, 3, 1, 1, 2, 1, 4] * 10
    })

# 1. Tentukan Quasi-Identifier
quasi_identifiers = ['Age at enrollment', 'Gender', 'Marital status']

# 2. Generalisasi 1: Umur
def generalize_age(age):
    if age < 20:
        return '<20'
    elif 20 <= age <= 24:
        return '20-24'
    elif 25 <= age <= 29:
        return '25-29'
    elif 30 <= age <= 34:
        return '30-34'
    else:
        return '35+'

df_proyek['kelompok_umur'] = df_proyek['Age at enrollment'].apply(generalize_age)

# 3. Generalisasi 2: Status Pernikahan
def generalize_marital(status):
    if status == 1:
        return 'Single'
    else:
        return 'Lainnya'

df_proyek['marital_grouped'] = df_proyek['Marital status'].apply(generalize_marital)

# 4. Hitung Nilai K setelah Generalisasi
quasi_identifiers_final = ['kelompok_umur', 'Gender', 'marital_grouped']

grouped_final = df_proyek.groupby(quasi_identifiers_final).size()

k_final = grouped_final.min()
total_groups = len(grouped_final)
groups_of_one = (grouped_final == 1).sum()

print("\nHASIL PENANGANAN K-ANONYMITY:")
print(f"Quasi-Identifiers: {quasi_identifiers_final}")
print(f"Nilai K akhir: {k_final}")
print(f"Total kelompok: {total_groups}")
print(f"Jumlah kelompok dengan 1 orang: {groups_of_one}")

if k_final >= 5:
    print("STATUS: BERHASIL mencapai target k >= 5.")
else:
    print("STATUS: GAGAL, butuh generalisasi lebih lanjut.")

# 5. Buang kolom asli yang sudah digeneralisasi
aman_df = df_proyek.drop(columns=['Age at enrollment', 'Marital status'])

# 6. Simpan hasil
# aman_df.to_csv('/content/drive/MyDrive/praktikum EDA/pertemuan3/dataset_aman.csv', index=False)
```
